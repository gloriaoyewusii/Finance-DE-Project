```python
## Close price is the column i've chosen

## I chose the 6th and 18th period because of my observation about the rise and fall of the bitcoin price from open to close
## such that, the during 18 hours, there's not much change, however during the 6 hours to the end of the day the close price 
## started showing signs of rise.

## However, looking at the tail end of the data, it was glaring that prices appeared volatile with fluctuations in
## the increase and decrease and the culmination of the close price at a steady decline. 

## I chose the Close price because it represents the final agreed trading price for each hourly candle. 
## Since the moving average is intended to summarize market behaviour over time, I wanted to use the completed 
## price for each period rather than the opening price or the intraperiod extremes (High and Low), which only 
## capture part of that hour's activity.

## During exploration of the hourly data, I observed that recent price movements often changed direction within a 
## relatively short number of hours. I therefore chose a 6-period Fast Moving Average to capture recent market 
## behaviour while using an 18-period Slow Moving Average to provide a broader reference trend. These values are 
## treated as initial design parameters and may be refined if the backtest shows they generate excessive
## or insufficient signals.


## Fast Moving Average

```

```python
import numpy as np
```

```python
conditions = [ micro_df["fma_6"].isna(),
              micro_df["sma_18"].isna(),
              micro_df["fma_6"] > micro_df["sma_18"],
              micro_df["fma_6"] < micro_df["sma_18"]
            ]

choices = [np.nan, np.nan, "FMA > SMA", "FMA < SMA", ]

micro_df["ma_relationship"] = np.select(conditions, choices, default='FMA == SMA')
```

```python
micro_df["ma_relationship"].tail(20)
```

```python
micro_df["prev_relationship"] = micro_df["ma_relationship"].shift(1)
```

```python
micro_df["crossover_event"] = micro_df["ma_relationship"] != micro_df["prev_relationship"]
```

```python
micro_df = micro_df.drop(columns=["crossover_event"])
```

```python
crossover_conditions = [
    micro_df["ma_relationship"].isna(),
    micro_df["prev_relationship"].isna(),
    micro_df["ma_relationship"] != micro_df["prev_relationship"]
    
]

crossover_choices = [np.nan, np.nan, True]

micro_df["crossover_event"] = np.select(crossover_conditions, crossover_choices, default='False')
```

```python
micro_df.head(100)
```

```python


signal_conditions = [
    micro_df["ma_relationship"].isna(),
    micro_df["prev_relationship"].isna(),
    (micro_df["prev_relationship"] == "FMA < SMA") & (micro_df["ma_relationship"] == "FMA > SMA"),
    (micro_df["prev_relationship"] == "FMA > SMA") & (micro_df["ma_relationship"] == "FMA < SMA")
]


signal_choices = [np.nan, np.nan, 'Buy', 'Sell']
micro_df["signal"] = np.select(signal_conditions, signal_choices, default='Hold')
```

```python
trade_conditions = [
    micro_df["signal"] == 'Buy',
    micro_df["signal"] == 'Sell',
    micro_df["signal"] == 'Hold'
]

trade_actions = ['Entry', 'Exit', 'None']



micro_df["trade_action"] = np.select(trade_conditions, trade_actions)
```

```python
# Assumption: At the beginning of the backtest, the strategy starts with no open BTC position (is_holding_btc = 0).
# This avoids relying on unknown trading activity before the dataset begins.
```

```python
is_holding_btc = 0

micro_df["is_holding_btc"] = 0

for index, signal in micro_df["signal"].items(): 
    if signal == 'Buy':
        is_holding_btc = 1        

    elif signal == 'Sell':
        is_holding_btc = 0

    micro_df.loc[index, "is_holding_btc"] = is_holding_btc
```

```python
trade_log = pd.DataFrame(columns=[
    "trade_id",
    "entry_time",
    "entry_price",
    "exit_time",
    "exit_price",
    "holding_duration_hours",
    "dollar_pnl",
    "percentage_return"
])
```

```python
micro_df["open_time"] = pd.to_datetime(micro_df["open_time"])
micro_df["close_time"] = pd.to_datetime(micro_df["close_time"])
```

```python
trade_id = 1

entry_time = None
entry_price = None


for index, row in micro_df.iterrows():

    signal = row["signal"]

    if signal == "Buy":

        entry_time = row["close_time"]
        entry_price = row["close"]

    elif signal == "Sell" and entry_time is not None:

        exit_time = row["close_time"]
        exit_price = row["close"]

        holding_duration = (
            exit_time - entry_time
        ).total_seconds() / 3600

        dollar_pnl = exit_price - entry_price

        percentage_return = (
            (exit_price - entry_price)
            / entry_price
        ) * 100

        trade_log.loc[len(trade_log)] = [
            trade_id,
            entry_time,
            entry_price,
            exit_time,
            exit_price,
            holding_duration,
            dollar_pnl,
            percentage_return
        ]

        trade_id += 1

        entry_time = None
        entry_price = None
```

```python
start_date = micro_df["close_time"].min()

print(start_date)


end_date = micro_df["close_time"].max()

print(end_date)

evaluation_period = end_date - start_date

print(evaluation_period)

total_days = evaluation_period.days

print(f"Total days: {total_days}")
```

```python
# Perormance Analysis

# 1. Total Number Of Trades
#This metric helps evaluate how active the strategy is during the evaluation period. 
# A larger number of completed trades provides more observations for evaluating the strategy's behaviour 
# and performance, while a very small number of trades may provide insufficient evidence to confidently judge 
# the strategy.


total_trades = len(trade_log)

print(f"Total trades: {total_trades}")
```

```python
# Trades per day

trades_per_day = (total_trades / total_days)

print(f"Trades per day: {trades_per_day:.2f}")
```

```python
# Trades per week

total_weeks = (total_days / 7)
print(f"Total weeks: {total_weeks}")

trades_per_week = total_trades / total_weeks
print(f"Total weekly trades: {trades_per_week}")
```

```python
#Trades per month

total_months = (total_weeks / 4)
print(f"Total months: {total_months}")

trades_per_month = total_trades / total_months
print(f"Total monthly trades: {trades_per_month}")
```

The strategy generated approximately 23 completed trades per month over the evaluation period. Relative to the seven-month backtesting period, this suggests moderate activity since it produced regular trading opportunities without entering or exiting positions too much. Additionally, the 161 completed trades is worth sampling and observing for evaluating the strategy's performance and risk characteristics.

```python
#2. Win Rate

total_win_count = (trade_log["dollar_pnl"] > 0).sum()

print(f"Total wins: {total_win_count}")

win_rate = ((total_win_count / total_trades) * 100).round(2)

print(f"Trades win rate: {win_rate}")
```

The moving average crossover strategy produced 63 winning trades out of 161 completed trades, resulting in a 39.13% win rate. While the strategy records more losing trades than winning trades, the win rate alone is not sufficient to assess its overall profitability. The profitability of the strategy depends on whether the gains from winning trades outweigh the losses from losing trades, and this will be examined using Profit Factor and Average Win/Loss.

```python
#3. Average & Median holding duration for winning trades

winning_trades = trade_log[trade_log["dollar_pnl"] > 0]

average_win_holding_duration = winning_trades["holding_duration_hours"].mean().round(2)

median_win_holding_duration = winning_trades["holding_duration_hours"].median()
```

The average (23.37 hours) and median (23.0 hours) holding durations for winning trades are very similar. This suggests that the winning trade durations are relatively consistent and are not heavily influenced by a few number of outliers or unusually long trades.

```python
#4. Loss Rate

total_loss_count = (trade_log["dollar_pnl"] < 0).sum()

print(f"Total losses: {total_loss_count}")

loss_rate = ((total_loss_count / total_trades) * 100).round(2)

print(f"Trades loss rate: {loss_rate}")

```

```python
#5. Average & Median holding duration for losing trades

losing_trades = trade_log[trade_log["dollar_pnl"] < 0]


average_loss_holding_duration = losing_trades["holding_duration_hours"].mean().round(2)


median_loss_holding_duration = losing_trades["holding_duration_hours"].median()


print("The average losing trade holding duration is: ", average_loss_holding_duration)
print("The median losing trade holding duration is: ", median_loss_holding_duration)
```

Although the strategy loses more often than it wins (61% losses), it appears to cut losing trades relatively quickly while allowing profitable trades to remain open about 3 times longer than the losing trades.

```python
#6. Profit Factor

total_profit = winning_trades["dollar_pnl"].sum().round(2)
print("Total profit: ", total_profit)

total_loss = losing_trades["dollar_pnl"].sum()
total_loss = abs(total_loss)
print("Total loss: ", total_loss)

profit_factor = (total_profit / total_loss).round(2)

print("Profit factor: ", profit_factor)
```

The strategy generated approximately 60,633 dollars in gross profits while incurring approximately $71,418 in gross losses, resulting in a Profit Factor of 0.85. Since the Profit Factor is below 1.0, the cumulative profits from winning trades were insufficient to offset the cumulative losses from losing trades over the evaluation period. This indicates that, despite allowing winning trades to remain open longer on average, the strategy did not produce enough total profit to achieve overall profitability during the backtest.

```python
#7. Average Win Amount

print(len(winning_trades))
average_win = (total_profit / len(winning_trades)).round(2)

print("Average win: ", average_win)
```

```python
#.8 Average Loss Amount

average_loss = (total_loss/len(losing_trades)).round(2)

print("Average loss: ", average_loss)
```

```python
#9. Buy and Hold Baseline Comparison

# Buy at the first available candle
buy_time = micro_df.iloc[0]["open_time"]
buy_price = micro_df.iloc[0]["open"]

# Sell at the last available candle
sell_time = micro_df.iloc[-1]["close_time"]
sell_price = micro_df.iloc[-1]["close"]

# Dollar Profit/Loss
buy_hold_dollar_pnl = sell_price - buy_price

# Percentage Return
buy_hold_return = ((sell_price - buy_price) / buy_price) * 100

# Holding Duration
buy_hold_duration = sell_time - buy_time

print(f"Buy Time: {buy_time}")
print(f"Buy Price: ${buy_price:,.2f}")

print(f"Sell Time: {sell_time}")
print(f"Sell Price: ${sell_price:,.2f}")

print(f"Dollar P&L: ${buy_hold_dollar_pnl:,.2f}")
print(f"Return: {buy_hold_return:.2f}%")
print(f"Holding Duration: {buy_hold_duration}")
```

```python
buy_hold_df = pd.DataFrame({
    "buy_time": [buy_time],
    "buy_price": [buy_price],
    "sell_time": [sell_time],
    "sell_price": [sell_price],
    "buy_hold_pnl": [buy_hold_dollar_pnl],
    "buy_hold_return": [buy_hold_return],
    "buy_hold_duration": [buy_hold_duration]
    })
```

```python
print(buy_hold_df)
```

```python
net_strategy_pnl = (total_profit - total_loss).round(2)

print("The moving average crossover strategy P&L: ", net_strategy_pnl)
print("The buy and hold P&L: ", buy_hold_df["buy_hold_pnl"])
```

The moving average crossover strategy did not achieve overall profitability during the evaluation period, as evidenced by a Profit Factor below 1 and a negative net return. However, compared with a passive Buy & Hold approach, the strategy experienced a substantially smaller loss, suggesting that its exit rules helped reduce exposure during a generally declining market.

```python
strategy_summary = pd.DataFrame({
    "strategy": ["Moving Average Crossover"],
    "total_trades": [total_trades],
    "win_rate": [win_rate],
    "loss_rate": [loss_rate],
    "total_profit": [total_profit],
    "total_loss": [total_loss],
    "net_pnl": [net_strategy_pnl],
    "profit_factor": [profit_factor],
    "average_win": [average_win],
    "average_loss": [average_loss],
    "average_win_duration": [average_win_holding_duration],
    "average_loss_duration": [average_loss_holding_duration],
})
```

```python
buy_hold_df["strategy"] = "Buy and Hold"
```

```python
performance_comparison = pd.concat( [strategy_summary, buy_hold_df],
    ignore_index=True)
```

```python
strategy_summary["risk_reward_ratio"] = (average_win / average_loss).round(2)

print(strategy_summary["risk_reward_ratio"])
```

The Risk-Reward Ratio indicates that the average profit earned from a winning trade is higher than the average loss incurred from a losing trade. In other words, each successful trade tends to generate more value than each unsuccessful trade loses. However, this advantage is undone by the relatively low win rate. Because losing trades occur much more frequently than winning trades, their cumulative effect outweighs the larger average gains from individual winning trades. This is reflected in the strategy's Profit Factor of 0.85 and its negative net profit.
